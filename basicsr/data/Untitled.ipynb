```python
import torch
import torch.nn as nn

class SharedKernelScerVoxelMixer(nn.Module):
    """
    입력:
      scer  : (B, 3, H, W)
      voxel : (B, 3, H, W)
    출력:
      scer_hat : (B, 3, H, W)

    동작(모든 (s,v)쌍 동시 처리, 커널은 1개 공유):
      diff_{s,v} = scer_s - voxel_v
      w_{s,v} = Conv2d( diff_{s,v} )  # (1->1) 커널 공유
      scer_hat_s = sum_v( w_{s,v} * voxel_v )
    """
    def __init__(self, kernel_size=3, bias=True):
        super().__init__()
        pad = kernel_size // 2
        # 공통 커널 1장: in=1, out=1 (모든 (s,v)쌍에 공유)
        self.kernel = nn.Conv2d(1, 1, kernel_size=kernel_size, padding=pad, bias=bias)

    def forward(self, scer: torch.Tensor, voxel: torch.Tensor) -> torch.Tensor:
        B, C, H, W = scer.shape
        assert C == 3 and voxel.shape[1] == 3, "입력 채널 수는 3이어야 합니다."

        # (1) 모든 (s,v)쌍의 차이를 한 번에 구성: (B, 3, 3, H, W)
        #     scer_s를 v축으로, voxel_v를 s축으로 브로드캐스트
        diff = scer.unsqueeze(2) - voxel.unsqueeze(1)  # (B, s=3, v=3, H, W)
        print("\ncheck scer.unsqueeze(2)\n", scer.unsqueeze(2).shape, "\n", scer.unsqueeze(2))
        print("\ncheck voxel.unsqueeze(1)\n", voxel.unsqueeze(1).shape, "\n", voxel.unsqueeze(1))
        print("\n////=======================////\n", diff, "\n////=======================////\n")

        # (2) 공통 커널 1장으로 모두 컨볼루션: (B*9, 1, H, W) -> (B*9, 1, H, W)
        diff_flat = diff.reshape(B * 9, 1, H, W).contiguous()
        w_flat = self.kernel(diff_flat)
        w = w_flat.view(B, 3, 3, H, W)  # (B, s, v, H, W)

        # (3) voxel_v를 s축으로 복제해서 곱 & v축 합산 → 채널별 누적
        voxel_exp = voxel.unsqueeze(1).expand(B, 3, 3, H, W)  # (B, s, v, H, W)
        scer_hat = (w * voxel_exp).sum(dim=2)  # (B, s=3, H, W)

        return scer_hat


def test_shared_kernel_mixer():
    torch.set_printoptions(sci_mode=False)

    # ----- 1) 입력 (B=1, C=3, H=W=2) -----
    scer = torch.tensor(
        [
            [  # C=3
                [[1.0, 2.0],
                 [3.0, 4.0]],

                [[-1.0, 0.5],
                 [2.0, -3.0]],

                [[0.0, -2.0],
                 [1.0,  3.0]],
            ]
        ]
    )  # shape: (1,3,2,2)

    voxel = torch.tensor(
        [
            [
                [[ 0.5, -1.0],
                 [ 2.0,  0.0]],

                [[-2.0,  1.5],
                 [ 0.0, -1.0]],

                [[ 1.0,  0.5],
                 [-1.0,  2.0]],
            ]
        ]
    )  # shape: (1,3,2,2)

    # ----- 2) 모델 구성 -----
    mixer = SharedKernelScerVoxelMixer(kernel_size=1, bias=True)

    # 커널을 w=1, b=0으로 설정 -> Conv(x) = x
    with torch.no_grad():
        mixer.kernel.weight.fill_(1.0)     # (out=1, in=1, 1, 1) = 1
        if mixer.kernel.bias is not None:
            mixer.kernel.bias.zero_()

    # ----- 3) forward -----
    out = mixer(scer, voxel)  # (1,3,2,2)

    # ----- 4) 기대값 계산 -----
    # 공식: scer_hat_s = sum_v ( (scer_s - voxel_v) * voxel_v )
    # 벡터화로 기대값 계산
    diff = scer.unsqueeze(2) - voxel.unsqueeze(1)          # (1, s=3, v=3, 2, 2)
    exp = (diff * voxel.unsqueeze(1)).sum(dim=2)           # (1, 3, 2, 2)

    # ----- 5) 비교/출력 -----
    print("=== Inputs ===")
    print("scer:\n", scer)
    print("voxel:\n", voxel)
    print("\n=== Output (model) ===")
    print(out)
    print("\n=== Expected (manual) ===")
    print(exp)

    # close check
    assert torch.allclose(out, exp, atol=1e-6), "Output != Expected!"

    print("\n[OK] 모델 출력이 기대값과 일치합니다.")

if __name__ == "__main__":
    test_shared_kernel_mixer()
```
