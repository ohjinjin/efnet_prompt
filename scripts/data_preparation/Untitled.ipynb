```python
import h5py
from pathlib import Path

H5_DIR = Path("/home/work/data/REBlur_rawevents/train")

def list_keys(h5_path, print_tree=False):
    def _print_tree(g, prefix=""):
        for k in g.keys():
            obj = g[k]
            if isinstance(obj, h5py.Group):
                print(f"{prefix}[G] {k}/")
                _print_tree(obj, prefix + "    ")
            else:
                shape = obj.shape if hasattr(obj, "shape") else None
                dtype = obj.dtype if hasattr(obj, "dtype") else None
                print(f"{prefix}[D] {k}  shape={shape}  dtype={dtype}")

    try:
        with h5py.File(h5_path, "r") as f:
            print(f"\n=== {h5_path.name} ===")
            if print_tree:
                _print_tree(f, prefix="")
            else:
                # 루트 바로 아래의 key만
                for k in f.keys():
                    obj = f[k]
                    if isinstance(obj, h5py.Group):
                        print(f"[G] {k}/")
                    else:
                        shape = obj.shape if hasattr(obj, "shape") else None
                        dtype = obj.dtype if hasattr(obj, "dtype") else None
                        print(f"[D] {k}  shape={shape}  dtype={dtype}")
    except OSError as e:
        print(f"[ERROR] {h5_path.name}: {e}")

def main():
    h5_files = sorted(H5_DIR.glob("*.h5"))
    if not h5_files:
        print(f"No .h5 files found in: {H5_DIR}")
        return
    for fp in h5_files:
        list_keys(fp, print_tree=False)  # 전체 트리를 보고 싶으면 True

if __name__ == "__main__":
    main()
```
